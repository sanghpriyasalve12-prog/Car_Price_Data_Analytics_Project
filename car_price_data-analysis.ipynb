# IQR outlier calculation (numeric columns)

```python
Q1=df1.quantile(0.25)
Q3=df1.quantile(0.75)

IQR = Q3- Q1

IQR
```

```python
(Q3 + 1.5*IQR)
```

```python
(Q1 - 1.5*IQR)
```

Explanation:

This calculates IQR bounds and removes rows having outliers in any numeric column.
Output meaning:

- IQR prints spread of middle 50%
- Upper/lower bounds show outlier cutoffs
- df_2 is the numeric-only cleaned version (outliers removed)

```python
df_2 = df1[~((df1 < (Q1-1.5*IQR)) | (df1 > (Q3+1.5*IQR))).any(axis=1)]  # masking 
```

# Price distribution (Histogram)

```python
print(df.columns)

plt.hist(df["Price"], bins=50)
plt.title("Distribution of Car Prices")
plt.xlabel("MSRP")
plt.ylabel("Number of Cars")
plt.show()
```

Explanation: 

Shows how prices are spread.

- Most cars cluster in mid/low price range

# HP vs Price (Scatter)

```python
plt.scatter(df["HP"], df["Price"])
plt.title("Engine Horsepower vs Car Price")
plt.xlabel("Engine HP")
plt.ylabel("MSRP")
plt.show()
```

Explanation: 

Tests relationship between engine power and price.

Output: 

Usually upward trend → higher HP tends to cost more.

# Highway MPG vs Price

# Average price by vehicle size

```python
df.groupby("Vehicle Size")["Price"].mean().plot(kind="bar")
plt.title("Average Price by Vehicle Size")
plt.ylabel("Average Price")
plt.show()
```

Explanation: 

Compares average price by size category.

Output: 

Bigger sizes often show higher average price.

# Average price by number of doors

```python
df.groupby("Number of Doors")["Price"].mean().plot(kind="bar")
plt.title("Average Price by Number of Doors")
plt.ylabel("Average MSRP")
plt.show()
```

Output: 

Bars show which door category costs more on average.

# Cars count by make (bar plot saved)

# Correlation sorted vs Price

```python
corr = df[["HP","Cylinders","MPG_C","MPG_h","Popularity","Price"]].corr()
print(corr["Price"].sort_values(ascending=False))
```

Explanation: 

Shows which numeric variables relate most with Price.

Output:
- Top positive number after Price itself = strongest positive driver
- Negative values = inverse relation (goes up, price goes down)

Typical result pattern:
- HP and Cylinders → positive
- MPG → often negative or weak
- Popularity → small positive/weak
